```python
import numpy as np
import pandas as pd
df=pd.read_csv('/content/tips.csv')
q1=df['total_bill'].quantile(0.25)
q3=df['total_bill'].quantile(0.75)
iqr=q3-q1

lower_limit=q1-1.5*iqr
upper_limit=q3+1.5*iqr
outliers=df[(df['total_bill']<lower_limit) | (df['total_bill']>upper_limit)]
outliers
```
